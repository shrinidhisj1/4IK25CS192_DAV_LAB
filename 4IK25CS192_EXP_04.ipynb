Lab Expirement 4 - Data Processing Handling Missing data using imputation (mean, median,forward fill) and encoding categorical variablea

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/student_performance_dataset (3).csv")
print("First 5 rows of dataset: ")
print(df.head())
```

```python
print("\nDataset Info:")
print(df.info())
```

```python
df.loc[0:5, 'Study_Hour_per_Week']=None
df.loc[10:15, 'Attendence_Rate']=None
df.loc[20:25, 'Past_Exam_Score']=None
df.loc[30:35, 'Final_Exam_Score']=None

print("\nMissing values after inserting sample Nans:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df['Study_Hour_per_Week']=df['Study_Hour_per_Week'].fillna(df['Study_Hour_per_Week'].mean())
```

```python
df['Attendence_Rate']=df['Attendence_Rate'].fillna(df['Attendence_Rate'].median())
```

```python
df['Past_Exam_Score']=df['Past_Exam_Score'].ffill()
```

```python
df['Final_Exam_Score']= df['Final_Exam_Score'].fillna(df['Final_Exam_Score'].mean())
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
df= pd.get_dumies(df, columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded columns:")
print(df.column.tolist())
```

```python
le = LabelEncoder()
df['Past_Fail_Encoder'] = le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Result:")
print(df[['Pass_Fail', 'Past_Fail_Encoder']].head())
```

```python
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Activities']=le.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv("student_performance_cleaned.csv",index=False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
