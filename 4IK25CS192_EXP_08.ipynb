Experiment-08 Logistic regression for binary classification

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay,accuracy_score, precision_score, recall_score, classification_report,f1_score )
import matplotlib.pyplot as plt
```

Load dataset

```python
df = pd.read_csv('/content/Titanic-Dataset.csv')
```

Preprocessing

```python
df = df.drop(['PassengerId','Name','Ticket','Cabin'], axis=1)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Sex'] = df['Sex'].map({'male':0,'female':1})
df['Embarked'] = df['Embarked'].map({'S':0,'C':1,'Q':2})
```

Features and Target

```python
x = df[['Pclass','Sex','Age','SibSp','Parch','Fare','Embarked']]
y = df['Survived']
```

Train Test Split (80-20)

```python
X_tr, X_te, y_tr, y_te = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

Model Building

```python
clf=LogisticRegression(max_iter=1000)
clf.fit(X_tr,y_tr)
y_pred=clf.predict(X_te)
```

Evaluation metrics

```python
print("Accuracy:",round(accuracy_score(y_te, y_pred),3))
print("Precision:",round(precision_score(y_te, y_pred),3))
print("Recall:",round(recall_score(y_te, y_pred),3))
print("F1-score:",round(f1_score(y_te, y_pred),3))
print('\n',classification_report(y_te, y_pred))
```

Confusion matrix

```python
ce=confusion_matrix(y_te,y_pred,labels=[0,1])
print("\nConfusionMatirx:\n",ce)
ConfusionMatrixDisplay(ce,display_labels=["Not Survived","Suvived"]).plot(cmap="Blues")
plt.title("Confusion Matrix -- Titanic Survival Prediction")
plt.show()
```
